```python
tickets = [
    ("My Wi-Fi is not working!", "Network"),
    ("I cannot connect to the internet.", "Network"),
    ("The network keeps disconnecting.", "Network"),

    ("I forgot my account password.", "Account"),
    ("I cannot login to my account.", "Account"),
    ("My account has been locked.", "Account"),

    ("My payment failed yesterday.", "Payment"),
    ("I was charged twice for my order.", "Payment"),
    ("My card payment is not going through.", "Payment"),

    ("My laptop keeps crashing.", "Hardware"),
    ("The keyboard is not working.", "Hardware"),
    ("My computer will not turn on.", "Hardware"),

    ("The application keeps freezing.", "Software"),
    ("The app crashes when I open it.", "Software"),
    ("I cannot install the software.", "Software"),
]
```

## Text Preprocessing

TODO(STUDENT-ML)


```python
class TextProcessor:
    def __init__(self, tickets):
        self.tickets = tickets
        



    def text_lower(self):
        return [(text.lower(), label) for text, label in self.tickets]



    def clean_text(self, text):
        return "".join(
            char for char in text 
            if char.isalnum() or char.isspace() or char in "-'"

    )



processor = TextProcessor(tickets)
lowered_tickets = processor.text_lower()
cleaned_tickets = lowered_tickets.clean_text()


for text, label in lowered_tickets:
    print(label, "->", text)




```
